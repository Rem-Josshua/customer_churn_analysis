```python
# Load the dataset and import the necessary libraries
import pandas as pd
pd.set_option('display.max_columns', 100)


df = pd.read_csv('WA_Fn-UseC_-Telco-Customer-Churn.csv')
df.info()
```

```python
#Remove duplicates, if there is any.
df = df.drop_duplicates()
```

```python
#Drop customerID column
df = df.drop('customerID', axis=1).copy()
```

```python
# rename columns tenure, and gender for more consistent naming:
df = df.rename(columns={'gender':'Gender', 'tenure':'Tenure'})
```

```python
# Change TotalCharges data type to the correct one
pd.to_numeric(df['TotalCharges'], errors='coerce').isnull().sum()

# TotalCharges column has 11 null values, let's investigate these null values.
```

```python
# Categorizing TotalCharges column:
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')
df.query('TotalCharges.isnull()')[['Tenure', 'TotalCharges']]

# Upon checking, null values in this column has 0 Tenure, hence no amount for TotalCharges yet.
```

```python
#Fill null values in TotalCharges column with 0
df['TotalCharges'] = df['TotalCharges'].fillna(0)
```

```python
df.to_csv('cleaned_data.csv', index=False)
```

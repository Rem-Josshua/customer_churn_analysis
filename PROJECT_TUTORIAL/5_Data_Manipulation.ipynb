```python
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt

dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

```python
df.pivot_table(values='salary_year_avg',index='job_title_short', aggfunc=['mean', 'median', 'std'])
```

```python
index_to_sort = df[['job_title_short', 'salary_year_avg']][df['job_title_short'] == 'Data Analyst']
index_to_sort.reset_index(inplace=True)
index_to_sort
```

```python
index_to_sort.set_index(keys='job_index', inplace=True)
```

```python
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt

dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

```python
#get the median to use for filling NA's
median_salary = df['salary_year_avg'].median()
median_salary_hour = df['salary_hour_avg'].median()
df_filled = df
```

```python
df_filled['salary_year_avg'] = df_filled['salary_year_avg'].fillna(value = median_salary)
df_filled['salary_hour_avg'] = df_filled['salary_hour_avg'].fillna(value = median_salary_hour)
```

```python
#drop duplicate values
df_unique = df_filled
df_unique = df_unique.drop_duplicates(subset=['job_title','company_name'])
```

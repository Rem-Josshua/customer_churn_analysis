```python
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt

dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

```python
#fill NaN values using the job median value # is it better to fill it with their respective medians rather than the universal median? ask chatgpt about this.
median_salary_year = df['salary_year_avg'].median()
median_salary_hour = df['salary_hour_avg'].median()
print(median_salary_year, median_salary_hour)
```

```python
df_filled = df
df_filled['salary_year_avg'] = df_filled['salary_year_avg'].fillna(median_salary_year)
df_filled['salary_hour_avg'] = df_filled['salary_hour_avg'].fillna(median_salary_hour)
```

```python
#drop duplicates to this new dataset
df_unique = df_filled
df_unique = df_filled.drop_duplicates()
#print data length after removing duplicates
print(len(df_filled) - len(df_unique))
```

```python
#drop duplicates using job_title and company_name as criteria to get rid of company's spamming job_title (usual doings of some company according to luke barousse).
df_unique = df_unique.drop_duplicates(subset = ['job_title', 'company_name'])
#print data length after removing duplicates
print('length of df_filled is: ',len(df_filled))
print('length of df_unique is: ',len(df_unique))
print('their difference is: ', len(df_filled) - len(df_unique))
```
